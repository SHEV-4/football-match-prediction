```python
name_dir = "17-23_02_26"
df_result = pd.read_csv(f"{name_dir}/whoscored_match_result.csv")
df_stats = pd.read_csv(f"{name_dir}/whoscored_combined.csv")
```

```python
def union_table(matches, stats, out_df, team_col, prefix, axis=0):
    frames = []

    for name in matches[team_col]:
        row_df = stats.loc[name].to_frame().T.add_prefix(prefix)
        frames.append(row_df)

    if not out_df.empty:
        frames.insert(0, out_df)

    return pd.concat(frames, ignore_index=True, axis=axis)
```

```python
def make_full_table(matches_df, stats_df):
    rows = []

    for _, row in matches_df.iterrows():
        # Статистика
        home_stats = stats_df.loc[row["home_team"]].add_prefix("home_")
        away_stats = stats_df.loc[row["away_team"]].add_prefix("away_")

        # Додаємо назву команди поряд зі своєю статистикою
        home_block = pd.concat([
            pd.Series({"home_team": row["home_team"]}),
            home_stats
        ])

        away_block = pd.concat([
            pd.Series({"away_team": row["away_team"]}),
            away_stats
        ])
        results_matches = pd.Series({"match_result":row["result_match"]})
        # Об’єднання у фінальний ряд
        full_row = pd.concat([home_block, away_block,results_matches])
        rows.append(full_row)

    return pd.DataFrame(rows).reset_index(drop=True)
```

```python
df = pd.DataFrame()
df = make_full_table(df_result,df_stats)
```

```python
df['home_team'] = df['home_team'].map(_TEAMS_ID)
df['away_team'] = df['away_team'].map(_TEAMS_ID)
df['home_league'] = df['home_league'].map(_LEAGUES_ID)
df['away_league'] = df['away_league'].map(_LEAGUES_ID)
```
